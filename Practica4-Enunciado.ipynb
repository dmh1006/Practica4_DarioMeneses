### Contexto
El entorno hospitalario se representa como una **matriz bidimensional** que modela zonas con diferentes penalizaciones de coste (pasillos, consultas, quirófanos, zonas restringidas, etc.).

Un hospital se representa como una matriz donde cada elemento representa un fragmento de mapa. Cada área tiene una penalización diferente (entrar en un quirófano que andar por el pasillo):
- `M`: obstáculo o muro (no transitable)
- `P`: Pasillo (penalización = 1)
- `D#`: despacho (penalización = 150)
- `C#`: consulta (penalización = 200)
- `UE#`: unidad de enfermería (penalización = 250)
- `B#`: baño (penalización = 20)
- `#N`: habitacion zona norte (penalización = 50)
- `#S`: habitacion zona sur (penalización = 50)
- `LB`: laboratorio (penalización = 500)
- `AP`: anatomía patológica (penalización = 1000)
- `I`: centro de información (penalización = 150)
- `F`: Farmacia (penalización = 2000)
- `Q#`: quirófano (penalización = 5000)
- `UCI#`: unidad de cuidados intensivos (penalización = 2500)
- `RX`: radiología (penalización = 1500)

### RECUERDA: 

Para poder obtener una visualización óptima de las ruta, se proporcionan las clases:
- `SolucionesPosiciones`: toma como parámetros para crear un objeto, el camino obtenido (ruta) como una colección indexada de posiciones (colección indexada de dos elementos) y el mapa del hospital.
- `VisualizadorHospital`: toma como argumento un objeto del tipo `SolucionesPosiciones` y el mapa del hospital. El objeto instanciado permite visualizar en html el mapa del hospital y la ruta incluida como solución, manejada mediante dos botones.

Un ejemplo de empleo de estas clases sería lo siguiente, siendo camino la ruta obtenida de un algoritmo de búsqueda:

```python
    soluciones_mostrar = SolucionesPosiciones(camino,mapa_hospital)
    visualizador_final = VisualizadorHospital(soluciones_mostrar,mapa_hospital)
    visualizador_final.mostrar()
```
